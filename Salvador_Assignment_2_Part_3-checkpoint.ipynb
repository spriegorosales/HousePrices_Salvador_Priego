```python
import pandas as pd
from matplotlib import pyplot as plt 
import seaborn as sns
```

```python
houses = pd.read_csv('houses.csv')
```

```python
houses.head()
```

```python
sns.jointplot(x = 'LotFrontage', y ='SalePrice', data = houses, kind = 'scatter', hue = 'YrSold', palette = 'Set2')
plt.title('Sale Price vs Lot Frontage', x = 0.5, y = 1.2)
plt.xlabel('Lot Frontage')
plt.ylabel("Sale Price")
plt.savefig('LotFrontage.jpg')
```

```python
sns.barplot(x = "MSZoning", y = "SalePrice", data = houses, hue = "MSZoning", palette = 'husl', legend = False)
plt.title('Sale Price vs Zoning Class', x = 0.5, y = 1)
plt.xlabel('Zoning Classification')
plt.ylabel("Sale Price")
plt.savefig('ZoningPlace.jpg')
```

```python
sns.jointplot(x = 'LotArea', y ='SalePrice', data = houses, kind = 'scatter', hue = 'YrSold', palette = 'Set2')
plt.title('Sale Price vs Lot Area', x = 0.5, y = 1.2)
plt.xlabel('Lot Area')
plt.ylabel("Sale Price")
#plt.savefig('LotArea.jpg')
```

```python
sns.barplot(x = "MSSubClass", y = "SalePrice", data = houses, hue = "MSSubClass", palette = 'husl', legend = False)
plt.title('Sale Price vs Building Class', x = 0.5, y = 1)
plt.xlabel('Building Class')
plt.ylabel("Sale Price")
plt.savefig('BuildingClass.jpg')
```

```python
sns.barplot(x = "Street", y = "SalePrice", data = houses, hue = "Street", palette = 'husl', legend = False)
plt.title('Sale Price vs Steet Type', x = 0.5, y = 1)
plt.xlabel('Street Type')
plt.ylabel("Sale Price")
plt.savefig('StreetType.jpg')
```

```python
sns.barplot(x = "Utilities", y = "SalePrice", data = houses, hue = "Utilities", palette = 'husl', legend = False)
plt.title('Sale Price vs Utilities', x = 0.5, y = 1)
plt.xlabel('Utilities')
plt.ylabel("Sale Price")
plt.savefig('Utilities.jpg')
```
